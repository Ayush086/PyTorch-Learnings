## Byte Pair Encoding Algorithm

```python
def get_stats(ids):
    counts = {}
    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1
    return counts

stats = get_stats(tokens)
print(sorted(((v, k) for k, v in stats.items()), reverse=True))
```

```python
top_pair = max(stats, key=stats.get)
top_pair
```

```python
def merge(ids, pair, idx):
    # in the list of ids, replace all consecutive occurences of pair with new token idx
    newids = []
    i = 0
    while i < len(ids):
        # if we're not at the very last position AND the pair matches, replace it
        if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
            newids.append(idx)
            i+= 2
        else:
            newids.append(ids[i])
            i+=1
    return newids

# for example
print(merge([5, 6, 6, 7, 9, 1], (6, 7), 99))
```

```python
vocab_size = 276 # desired size
num_merges = vocab_size - 256
ids = list(tokens)

merges = {} #(int, int) --> int
for i in range(num_merges):
    stats = get_stats(ids)
    pair = max(stats, key=stats.get)
    idx = 256 + i
    print(f"merging {pair} into a new token {idx}")
    ids = merge(ids, pair, idx)
    merges[pair] = idx
```

```python
# before and after scenario
print("tokens length: ", len(tokens))
print('ids length: ', len(ids))
print(f"compression ratio: {len(tokens)/len(ids):.2f}x")
```

now let's see how we can decode this encoded token

```python
vocab = {idx: bytes([idx]) for idx in range(256)}
for(p0, p1), idx in merges.items():
    vocab[idx] = vocab[p0] + vocab[p1]
    
def decode(ids):
    tokens = b"".join(vocab[idx] for idx in ids)
    text = tokens.decode("utf-8", errors="replace")
    return text

print(decode([128]))
```

```python
def encode(text):
    tokens = list(text.encode('utf-8'))
    while len(tokens) >= 2:
        stats = get_stats(tokens)
        pair = min(stats, key=lambda p: merges.get(p, float("inf")))
        if pair not in merges:
            break # all merges covered
        idx = merges[pair]
        tokens = merge(tokens, pair, idx)
    return tokens

print(encode("new learning"))
```
